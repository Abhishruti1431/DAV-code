# Exploratory Data Analysis - Iris Dataset

This notebook performs basic exploratory data analysis on the Iris flower dataset using Pandas, Matplotlib and Seaborn.

The analysis includes:
- Loading and inspecting the dataset
- Checking columns and basic statistics
- Scatter plots
- Bar plots
- Histograms
- Box plots
- Pair plots
- Correlation matrix
- Correlation heatmap


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Loading the Dataset

```python
from sklearn.datasets import load_iris

iris = load_iris()

data = pd.DataFrame(
    iris.data,
    columns=["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
)

data["Species"] = pd.Categorical.from_codes(
    iris.target,
    iris.target_names
)

data.head()
```

## Scatter Plots

```python
plt.scatter(data["SepalLengthCm"], data["SepalWidthCm"])

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Sepal Length vs Sepal Width")

plt.show()
```

```python
plt.scatter(data["PetalLengthCm"], data["PetalWidthCm"])

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Petal Length vs Petal Width")

plt.show()
```

```python
plt.scatter(data["SepalLengthCm"], data["PetalLengthCm"])

plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.title("Sepal Length vs Petal Length")

plt.show()
```

```python
plt.scatter(data["SepalWidthCm"], data["PetalWidthCm"])

plt.xlabel("Sepal Width")
plt.ylabel("Petal Width")
plt.title("Sepal Width vs Petal Width")

plt.show()
```

```python
plt.scatter(data["SepalLengthCm"], data["PetalWidthCm"])

plt.xlabel("Sepal Length")
plt.ylabel("Petal Width")
plt.title("Sepal Length vs Petal Width")

plt.show()
```

```python
plt.scatter(data["PetalLengthCm"], data["SepalWidthCm"])

plt.xlabel("Petal Length")
plt.ylabel("Sepal Width")
plt.title("Petal Length vs Sepal Width")

plt.show()
```

## Species Distribution

```python
data["Species"].value_counts().plot(kind="bar")

plt.xlabel("Species")
plt.ylabel("Count")
plt.title("Number of Flowers in Each Species")

plt.show()
```

## Histograms

```python
plt.hist(data["SepalLengthCm"])

plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.title("Sepal Length Distribution")

plt.show()
```

```python
plt.hist(data["SepalWidthCm"])

plt.xlabel("Sepal Width")
plt.ylabel("Frequency")
plt.title("Sepal Width Distribution")

plt.show()
```

```python
plt.hist(data["PetalLengthCm"])

plt.xlabel("Petal Length")
plt.ylabel("Frequency")
plt.title("Petal Length Distribution")

plt.show()
```

```python
plt.hist(data["PetalWidthCm"])

plt.xlabel("Petal Width")
plt.ylabel("Frequency")
plt.title("Petal Width Distribution")

plt.show()
```

```python
data[
    ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
].hist(bins=10, figsize=(10, 8))

plt.show()
```

## Box Plots

```python
data[
    ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
].boxplot()

plt.title("Iris Dataset Measurements")
plt.ylabel("Values")
plt.show()
```

```python
sns.boxplot(
    data=data[
        ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
    ]
)

plt.title("All Features Box Plot")
plt.show()
```

```python
sns.boxplot(x="Species", y="PetalLengthCm", data=data)

plt.title("Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")
plt.show()
```

```python
sns.boxplot(x="Species", y="PetalWidthCm", data=data)

plt.title("Petal Width by Species")
plt.xlabel("Species")
plt.ylabel("Petal Width")
plt.show()
```

```python
sns.boxplot(x="Species", y="SepalLengthCm", data=data)

plt.title("Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length")
plt.show()
```

## Pair Plot

```python
sns.pairplot(data, hue="Species")

plt.show()
```

## Correlation Matrix

```python
data.corr(numeric_only=True)
```

```python
correlation = data.corr(numeric_only=True)

sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")
plt.show()
```

```python
corr = data[
    ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
].corr()

sns.heatmap(corr, annot=True)

plt.title("Correlation Matrix")
plt.show()
```

## Observations

1. Petal length and petal width show a strong positive relationship.
2. Sepal length also has a positive relationship with petal length and petal width.
3. The three Iris species have noticeably different petal measurements.
4. The bar plot shows that all three species contain the same number of observations.
5. The histograms show the distribution of the four numerical flower measurements.
6. The box plots help identify the spread and possible outliers in the measurements.
7. The pair plot shows that petal-related measurements are useful for distinguishing between the different species.


## Conclusion

The Iris dataset contains four numerical measurements and one categorical variable, Species. Basic exploratory data analysis shows clear relationships between the measurements, especially between petal length and petal width. The visualizations also show that the species differ considerably in their petal measurements.
