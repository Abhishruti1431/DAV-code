```python
import numpy as np
import pandas as pd

marks=np.array([44,55,67,21,93])
print("Original Array:")
print(marks)

print("---NumPy Indexing---")
print("First mark:", marks[0])
print("Last mark:", marks[-1])
print("second element:", marks[1])

print("---NumPy Slicing---")
print("marks from start to third element:", marks[0:3])
print("print elements in reverse order:", marks[::-1])
print("print every second element:", marks[::2])
print("print elements from inde 2 to 4", marks[2:5])


```
