```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_wine

wine = load_wine()

data = pd.DataFrame(
    wine.data,
    columns=wine.feature_names
)

data["WineClass"] = wine.target
```
